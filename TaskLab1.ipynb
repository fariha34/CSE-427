Numphy Tasks

```python
#Task 1
a=np.random.randint(1,101,size=20)
b=np.random.randint(1,101,size=20)
print(f'a={a}')
print(f'b={b}')
```

```python
#Task 2
arr1=a.reshape(5,4)
arr2=b.reshape(5,4)
print(f"Matrix a")
print(arr1)
print(f"Matrix b")
print(arr2)
```

```python
#Task 3 -scalar multi
print(arr1*arr2)
```

```python
#Task4
arr2 = arr2.T
print(f"Changed matrix b")
print(arr2)
c = arr1 @ arr2
print(f"Matrix Multiplication")
print(c)
```

```python
#Task 5
print(c[:,2:4])
print(c[0:5,0:5])
```

```python
#Task 6
print(np.max(c))
print(np.argmax(c))
```

```python
#Task 7
arr3=c.flatten()
print(f"1D Array")
print(arr3)
print(f"Original")
print(c)

print(arr3.ravel())
```

Matplotlib Tasks

```python
#Task 1
temperature_dhaka = [25,34,21,45,28,6,43,18,7,2]
humidity_dhaka = [28, 25,29,20, 26, 50, 19, 29, 52, 55]
plt.scatter(temperature_dhaka,humidity_dhaka,marker="*",color="r")
plt.title("Temperature vs Humidity")
plt.xlabel("Temperature")
plt.ylabel("Humidity")
plt.show()
```

```python
#Task 2
study_hours = [2,3,4,4, 5, 6, 7, 7, 8, 9, 9, 10, 11, 11, 12]
marks = [6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100]
plt.figure(figsize=(12,8))
plt.plot(study_hours,marks)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
#Task 3
subjects = ['Maths', 'English', 'Science', 'Physics', 'Computer']
marks_s = [89, 90, 45, 78, 99]
plt.barh(subjects,marks_s)
plt.title("Subject Scores Horizontal")
plt.xlabel("Marks")
plt.show()
```

```python
#Task 4
colors = ["red","g","b","orange","pink"]
plt.bar(subjects, marks_s, color=colors)
plt.title("Subject Scores")
plt.ylabel("Marks")
plt.show()
```

```python
#Task 5
plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.scatter(temperature_dhaka,humidity_dhaka,marker="*",color="r")
plt.title("Temperature vs Humidity")

plt.subplot(2,2,2)
plt.plot(study_hours,marks)
plt.title("Study Hours vs Marks")

plt.subplot(2,2,3)
plt.barh(subjects,marks_s)
plt.title("Subject Scores Horizontal")

plt.subplot(2,2,4)
plt.bar(subjects,marks_s,color=colors)
plt.title("Subject Scores Vertical")

plt.tight_layout()
plt.show()
```
